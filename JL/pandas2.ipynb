# Read Excel Files with Pandas using read_excel()

```python
import pandas as pd
import matplotlib.pyplot as plt
```

# read excel file into pandas DataFrame

```python
df =pd.read_excel("test.xlsx")
```

# display DataFrame

```python
print(df)
```

# display DataFrame in graphic

```python
df.plot()
plt.show()
```

# display DataFrame in graphic in bar

```python
df.plot.bar(stacked=True)
plt.show()
```

# other graphic

```python
plt.plot(df['x'],df['x*2'],df['x*3'])
plt.show()
```
